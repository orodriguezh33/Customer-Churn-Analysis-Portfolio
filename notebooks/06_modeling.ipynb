# 06 - Modeling

Goal: train candidate models on the same encoded dataset to compare them fairly in the next notebook.

```python
import os

import joblib
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

PROCESSED_DIR = "../data/processed"
MODELS_DIR = "../models"

train_encoded = pd.read_parquet(f"{PROCESSED_DIR}/train_encoded.parquet")
X_train = train_encoded.drop(columns=["churn_flag"])
y_train = train_encoded["churn_flag"]

print(f"X_train: {X_train.shape[0]} rows x {X_train.shape[1]} columns")
```

## Candidate models

Logistic Regression, Decision Tree, Random Forest, and XGBoost are trained.

Logistic Regression uses `StandardScaler` inside a `Pipeline`. Tree-based models don't require scaling.

```python
models = {
    "logistic_regression": Pipeline([
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(max_iter=1000, random_state=42)),
    ]),
    "decision_tree": DecisionTreeClassifier(random_state=42),
    "random_forest": RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    "xgboost": XGBClassifier(random_state=42, eval_metric="logloss"),
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name}: trained")
```

## Save the trained models

```python
os.makedirs(MODELS_DIR, exist_ok=True)

for name, model in models.items():
    joblib.dump(model, f"{MODELS_DIR}/{name}.joblib")

print(f"Saved: {[f'{name}.joblib' for name in models]}")
```
